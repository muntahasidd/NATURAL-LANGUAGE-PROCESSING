**Data Cleaning**

```python
#REMOVE HTML TAGS
def remove_html(text):
 return re.sub(r'<.*?>', '', text)

#REMOVE URLs
def remove_urls(text):
 return re.sub(r'http\S+|www.\S+', '', text)

#REMOVE PUNCTUATION
def remove_punctuations(text):
 return text.translate(str.maketrans('', '', string.punctuation))

#REMOVE NUMBERS
def remove_numbers(text):
 return re.sub(r'\d+', '', text)

#REMOVE STOPWORDS
def remove_stopwords(text):
 stop_words = set(stopwords.words('english'))
 return ' '.join([word for word in text.split() if word.lower() not in stop_words])

#REMOVE EMOJIS
def replace_emojis(text):
 return emoji.demojize(text)

#CORRECT SLANGS
def correct_slang(text):
 slang_dict = {
"omg": "oh my god",
"rn": "right now",
"soooo": "so",
"looove": "love",
"lit": "amazing",
"tho": "though",
"af": "very",
"im": "i am"
}
 for slang, full in slang_dict.items():
  text = re.sub(r'\b' + slang + r'\b', full, text, flags=re.IGNORECASE)
  return text






```

**Apply Preprocessing Steps**

```python
def clean_text(text):
 text = remove_html(text)
 text = remove_urls(text)
 text = remove_punctuations(text)
 text = remove_numbers(text)
 text = remove_stopwords(text)
 text = replace_emojis(text)
 text = correct_slang(text)
 return text




df['cleaned'] = df['comment'].apply(clean_text)
print("\nCleaned Text:\n", df[['user', 'cleaned']])
```

**Tokenization, Stemming, Lemmatization**

```python
#TOKENIZATION
def tokenize_text(text):
 return word_tokenize(text)

#STEMMING
def stem_tokens(tokens):
 stemmer = PorterStemmer()
 return [stemmer.stem(t) for t in tokens]

#LEMMATIZATION
def lemmatize_tokens(tokens):
  lemmatizer = WordNetLemmatizer()
  return [lemmatizer.lemmatize(t) for t in tokens]
```

```python
nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
```

```python
df['tokens'] = df['cleaned'].apply(tokenize_text)
df['stemmed'] = df['tokens'].apply(stem_tokens)
df['lemmatized'] = df['tokens'].apply(lemmatize_tokens)
```

**DISPLAY OUTPUTS**
